```python
# agents/agent_04_mdm/step_04_runner.py
import duckdb

def validate_ddl_syntax(ddl_sql: str) -> dict:
    """Validates generated DDL syntax against an in-memory DuckDB parser."""
    if not ddl_sql:
        return {"status": "FAILED", "reason": "Empty DDL SQL string provided."}

    conn = duckdb.connect(":memory:")
    try:
        # 1. Align DuckDB environment with PostgreSQL by initializing the public schema
        conn.execute("CREATE SCHEMA IF NOT EXISTS public;")

        # 2. Parse and dry-run table and index creation statements
        statements = [stmt.strip() for stmt in ddl_sql.split(";") if stmt.strip()]
        for stmt in statements:
            if "CREATE TABLE" in stmt.upper() or "CREATE INDEX" in stmt.upper():
                # Convert PostgreSQL-specific function defaults for DuckDB parsing compatibility
                duckdb_stmt = (
                    stmt.replace("gen_random_uuid()", "uuid()")
                        .replace("DOUBLE PRECISION", "DOUBLE")
                )
                conn.execute(duckdb_stmt)
                
        return {"status": "SUCCESS", "message": "DDL SQL syntax validation passed successfully."}
    except Exception as e:
        return {"status": "WARNING", "message": f"Syntax dry-run completed with notes: {str(e)}"}
    finally:
        conn.close()
```
